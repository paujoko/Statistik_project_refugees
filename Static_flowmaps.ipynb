# Visualisierung globaler Flüchtlingsströme

Dieses Notebook dient der grafischen Darstellung der zuvor aufbereiteten UNHCR-Fluchtbewegungsdaten.
Es lädt die aggregierten Flussdaten mit geografischen Koordinaten und visualisiert sie als Flowmaps auf einer Weltkarte.
Dabei werden globale Ströme, kontinentale Muster sowie Zu- und Abwanderungsbewegungen einzelner Länder für ein ausgewähltes Jahr dargestellt.

```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.patches as mpatches

flows = pd.read_csv(
    "output_csv_files/flows_with_coords_2016.csv"
)

world = gpd.read_file(
    "output_csv_files/world_with_centroids.geojson"
)

YEAR = 2016

print("Flows:", flows.shape)
flows.head()
```

Diese Funktion visualisiert Fluchtbewegungen als Pfeile auf einer Weltkarte, wobei nur die stärksten Ströme berücksichtigt und visuell hervorgehoben werden.  
Die Flüsse werden nach ihrer Größe logarithmisch klassifiziert, farblich abgestuft und in sinnvoller Zeichenreihenfolge geplottet, um Überlagerungen zu reduzieren und die Lesbarkeit zu verbessern.

```python
def plot_flow_map(world, flows, year, title, max_flows=300):
    """
    Flowmap mit:
    - logarithmischen Bins
    - klaren Rottönen
    - großen Strömen im Vordergrund
    - robuster Behandlung von Problemfällen (kein StopIteration)
    """

    flows = flows.sort_values("Value", ascending=False).head(max_flows).copy()

    flows["Value"] = pd.to_numeric(flows["Value"], errors="coerce")

    flows = flows.dropna(
        subset=["Value", "origin_lon", "origin_lat", "asylum_lon", "asylum_lat"]
    )

    flows = flows[
        (flows["origin_lon"] != flows["asylum_lon"]) |
        (flows["origin_lat"] != flows["asylum_lat"])
    ]

    if flows.empty:
        print("Keine gültigen Flows zum Plotten.")
        return

    values = flows["Value"].to_numpy()

    vmin, vmax = values.min(), values.max()

    if vmin <= 0 or np.isnan(vmin):
        vmin = 1
    if vmax <= vmin:
        vmax = vmin + 1

    log_bins = np.logspace(np.log10(vmin), np.log10(vmax), 6)

    colors = ["#ff9999", "#ff6666", "#ff3333", "#cc0000", "#990000"]

    cat_idx = np.digitize(values, log_bins[1:-1])
    cat_idx = np.clip(cat_idx, 0, 4)
    flows["color_cat"] = cat_idx

    flows = flows.sort_values("color_cat", ascending=True)

    fig, ax = plt.subplots(figsize=(22, 12))

    world.plot(ax=ax, color="#E6E6E6", edgecolor="#BBBBBB", linewidth=0.5)

    for _, row in flows.iterrows():
        cat = int(row.color_cat)
        color = colors[cat]
        linewidth = 0.8 + 3.5 * ((cat + 1) / 5)

        ax.annotate(
            "",
            xy=(row.asylum_lon, row.asylum_lat),
            xytext=(row.origin_lon, row.origin_lat),
            arrowprops=dict(
                arrowstyle="-|>",
                color=color,
                lw=linewidth,
                shrinkA=0,
                shrinkB=0
            )
        )

    ax.set_title(f"{title} {year}", fontsize=26, pad=20)
    ax.set_xlabel("Longitude")
    ax.set_ylabel("Latitude")
    ax.grid(color="#DDDDDD", linestyle="--", linewidth=0.3)

    plt.show()
```

Dieser Funktionsaufruf erstellt eine Weltkarte, auf der die globalen Flüchtlingsströme für 2016 dargestellt werden, wobei nur die 1000 grössten Flüchtlingsströme geplottet werden.

```python
plot_flow_map(
    world,
    flows,
    YEAR,
    title="Globale Flüchtlingsströme",
    max_flows=1000
)
```

Dieser Abschnitt erstellt eine Zuordnung von Ländern zu Kontinenten auf Basis der Welt-Geodaten und ergänzt die Flussdaten um die Kontinente der Herkunfts- und Aufnahmeländer, um regionale Analysen zu ermöglichen.

```python
COUNTRY_TO_CONTINENT = dict(zip(world["country"], world["continent"]))

def get_continent(country):
    return COUNTRY_TO_CONTINENT.get(country, None)

flows["origin_continent"] = flows["Origin_clean"].apply(get_continent)
flows["asylum_continent"] = flows["Asylum_clean"].apply(get_continent)
```

Diese Funktion erstellt für jeden Kontinent separate Karten, die ausschliesslich die von dort ausgehenden Flüchtlingsströme darstellen, um regionale Herkunftsmuster besser vergleichen zu können.

```python
def plot_all_continents(world, flows, year, max_flows=300):

    continents = flows["origin_continent"].dropna().unique()

    for cont in continents:
        subset = flows[flows["origin_continent"] == cont]

        if subset.empty:
            continue

        print(f"Plot {cont} ({len(subset)} Flows)")
        plot_flow_map(
            world,
            subset,
            year,
            title=f"Refugee Flows from {cont}",
            max_flows=max_flows
        )
```

```python
plot_all_continents(world, flows, YEAR, max_flows=300)
```

Diese Funktion erstellt für jeden Kontinent separate Karten, die ausschliesslich die von dort eingehenden Flüchtlingsströme darstellen, um regionale Herkunftsmuster besser vergleichen zu können.

```python
def plot_all_continents_inbound(world, flows, year, max_flows=300):

    continents = flows["asylum_continent"].dropna().unique()

    for cont in continents:
        subset = flows[flows["asylum_continent"] == cont]

        if subset.empty:
            continue

        print(f"Inbound {cont} ({len(subset)} Flows)")
        plot_flow_map(
            world,
            subset,
            year,
            title=f"Refugee Inflows into {cont}",
            max_flows=max_flows
        )
```

```python
# %%
plot_all_continents_inbound(world, flows, YEAR, max_flows=300)
```

Diese Funktion erzeugt separate Karten für die ausgehenden und eingehenden Flüchtlingsströme eines ausgewählten Landes in einem Jahr, sodass Herkunfts- und Aufnahmerollen klar getrennt analysiert werden können.








```python
# %%
def plot_country_flows(world, flows, year, country, max_flows=300):

    flows_from = flows[flows["Origin_clean"] == country]
    flows_into = flows[flows["Asylum_clean"] == country]

    if not flows_from.empty:
        plot_flow_map(
            world,
            flows_from,
            year,
            title=f"Refugee Flows FROM {country}",
            max_flows=max_flows
        )

    if not flows_into.empty:
        plot_flow_map(
            world,
            flows_into,
            year,
            title=f"Refugee Flows INTO {country}",
            max_flows=max_flows
        )
```

```python
# %%
plot_country_flows(
    world,
    flows,
    YEAR,
    country="Afghanistan",
    max_flows=20
)
```
